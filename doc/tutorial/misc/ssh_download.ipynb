# SSH downloading

This notebook allows to download experiments directly from the spectrometer using ssh connection

##### Imports

```python
import os
import sys
from warnings import warn
from pprint import pprint

from masai.core.tools.ssh import SSHSession
```

### Connection to the AVANCE 400 or 500 at LCS

##### I/O directories

```python
DISTANTDIR = os.path.join('data','siblani','nmr')
LOCALDIR = os.path.expanduser(os.path.join('~','Dropbox','storage','bruker'))   # just above ../data
```

##### Establish a connection 
Of course one needs to have access to the spectrometer worstation (i.e., using *VPN*, when outside the laboratory)

```python
AVANCE400 = True
if AVANCE400:
    session = SSHSession("nmr", "192.168.223.175", verbose=True)
else:
    session = SSHSession("nmr", "192.168.223.185", verbose=True)
if not session.connected:
    warn('not connected')
else:
    print 'connected',session
```

##### Listing on the server
We are now going to execute a script on server to get all available experiment in a given directory


```python
%%file script_listing.py
"""This listing is gonna be uploaded to the server
"""
import os
import sys
args = sys.argv
path = args[1]
catalog = {}
count = 0
for root, dirs, files in os.walk(path):
    for fic in files:
        if fic in ['ser','fid']:
            name, expno = os.path.split(root)
            if name not in catalog.keys():
                catalog[name]=[]
            # add some information
            catalog[name].append(expno)
            count += 1
            
if count == 0:
    print 'no file found'

print catalog
```

The following file include the data we wants to exclude from downloading

```python
%%file exclude.txt
1r
1i
2rr
2ri
2ir
2ii
outd
thumb.png
shimvalues
specpar
uxnmr.*
scon2
*.temp
```

```python
file_ = 'script_listing.py'
session.scp(file_, 'to', file_)

fileexcl_ = 'exclude.txt'
session.scp(fileexcl_, 'to', fileexcl_)

_catalog = session.ssh('python script_listing.py %s'%DISTANTDIR)

catalog = eval(_catalog)

catalog.keys()
```

```python
import tarfile

# put here the keys you want to download
keys_selected = ['data/siblani/nmr/polymere encapsule 2',]  

# or if we wants all available keys, uncomment the following line
#keys_selected = catalog.keys()

# the same for the expnos
# expnos = 'all'
# or uncomment here
expnos =[501, 511, 516, 601, 611, 616, 500, 510, 515, 600, 610, 615]

for key in keys_selected:
    if expnos == 'all':
        expnos = catalog[key]
    for item in expnos:
        p = os.path.join(LOCALDIR,key,str(item))
        if not os.path.exists(p): 
            f= os.path.join(key,str(item))
            print "%s does not exist -> download."%p
            session.ssh("rm tempo.tar.gz; tar cvzf tempo.tar.gz -X exclude.txt '%s'"%(f.strip()))
            session.scp(LOCALDIR,"from","tempo.tar.gz")
            try:
                tar = tarfile.open(os.path.join(LOCALDIR,"tempo.tar.gz"))
                tar.extractall(path=LOCALDIR)
                tar.close()
            except:
                print 'error: bypassed'
        else:
            print "%s already exist. Bypass"%p
```
