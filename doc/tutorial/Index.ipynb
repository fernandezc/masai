## What is this tutorial for?

In this tutorial, we provide some examples on how to use <span style="color:orange;">***MASAÏ***</span> for reading and processing NMR (Bruker) files

The available tutorial parts are just here:

* [Part_1](./Part_1.basics.ipynb): Basics of 1D spectrum processing and plotting
* [Part_2](./Part_2.fitting.ipynb): 1D NMR spectra fitting
* [Part_3](./Part_3.series.ipynb): Handling series of 1D NMR spectra
* [Part_4](./Part_4.2D.ipynb): Processing and plotting 2D NMR spectrum
* [Part_5](./Part_5.ST_or_MQMAS.ipynb): Processing STMAS or MQMAS spectra
* [Part_6](./Part_6.Multiplex.ipynb): Processing Multiplex acquired MQMAS or STMAS spectra
* [Part_7](./Part_7.inadequate.ipynb): Processing and analysing INADEQUATE (or 2Q) spectra
* [Part_8](./Part_8.EXSY): Processing and analysing EXSY spectra
