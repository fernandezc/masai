# Appendice 1. Sources

`Source` class is the basic data container.

Plugins such as `Bruker` are derived from this class.

Here are some information about the propertie sof this class.

```python
from __future__ import division, print_function
```

```python
from masai.core.source import Source
import numpy as np
import pandas as pd
```

```python
# create a simple 1D source
source1d = Source()
x = np.arange(100.)/50.
y = np.sin(x*np.pi)
source1d.data = pd.DataFrame([y], columns=x)
source1d
```

```python
# get some information
source1d.size, source1d.shape, source1d.dtype
```

```python
# make a copy
source1d_copy = source1d.copy()
source1d_copy
```

```python
source1d_copy = source1d.copy()
# should be different object
assert(source1d_copy is not source1d)
# but equal contents
assert(source1d_copy == source1d)
```

```python
source1d_copy = source1d.copy()
# modify sourceb
source1d_copy /= 2
# now should be different, right?
assert(source1d_copy != source1d)
```

```python
# slicing on index (stop index not included)
source1d[10:14]
```

```python
# slicing on location (take the closest existing location in the source)
source1d.loc[0.21:0.35]
```

```python
# get  a single item by index
source1d[10]
```

```python
# get  a single item by location (the closest location)
source1d.loc[0.685]
```

```python
# create a simple 2D source (of complex type)
source2d = Source()
y = np.arange(20.)/20.
z = np.array([[np.exp(xi*2j*np.pi)*np.exp(yi*2j*np.pi) for xi in x] for yi in y])
source2d.data = pd.DataFrame(z, index=y, columns=x )
source2d
```

```python
# on the last dimension should work as for 1d
# slicing on columns
source2d[:,10:14]
```

```python
source1d.loc[0.21:0.35]
```

```python
# slicing on location
s1 = source1d.loc[0.21:0.35]
s2 = source2d.loc[:,0.21:0.35]
assert(np.all(s1.data.columns == s2.data.columns))
s2
```

```python
# get  a single item by index
source2d[:,10]
```

```python
# get  a single item by location (the closest)
source2d.loc[0.15,1.9]
```
