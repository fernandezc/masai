```python
def plot_graph(G, fig=None, subplot=111, figsize=(4,4), layout='circular',  disp='weight'):
        
    nunself = G.nodes()
    nself3 = []
    nself2 = []
    nself1 = []
    everysmall = []
    
    if disp=='weight':
        
        elarge=[(u,v,d) for (u,v,d) in G.edges(data=True) 
            if d['weight']>1 and u!=v]
        esmall=[(u,v,d) for (u,v,d) in G.edges(data=True) 
            if d['weight']==1]

        for (u,v,d) in G.edges(data=True):
            if u==v and d['weight']==3:
                nself3.append(u)
                nunself.remove(u)
            elif u==v and d['weight']==2:
                nself2.append(u)
                nunself.remove(u)
            elif u==v and d['weight']==1:
                nself1.append(u) 
                nunself.remove(u)
    else:
        mx = 0
        for (u,v,d) in G.edges(data=True):
            mx = max(mx, G[u][v][disp])
        print "mx", mx
        
        elarge=[(u,v,d) for (u,v,d) in G.edges(data=True) 
            if d[disp]>=2.*mx/3. and u!=v]
        esmall=[(u,v,d) for (u,v,d) in G.edges(data=True) 
            if (d[disp]>=mx/3. and d[disp]<mx/2.)]
        everysmall=[(u,v,d) for (u,v,d) in G.edges(data=True) 
            if ( d[disp]>0 and d[disp]<mx/4.)]
        
        for (u,v,d) in G.edges(data=True):
            if u==v and d[disp]>=mx/2.:
                nself3.append(u)
                nunself.remove(u)
            elif u==v and (d[disp]>=mx/4. and d[disp]<mx/2.):
                nself2.append(u)
                nunself.remove(u)
            elif u==v and (d[disp]>0 and d[disp]<mx/4.):
                nself1.append(u) 
                nunself.remove(u)
            
    #print nunself, nself1, nself2, nself3
    if not fig:
        fig = pl.figure(figsize=figsize)
    
    ax = fig.add_subplot(subplot)
    
    if layout == 'spring':
        pos=nx.spring_layout(G) # positions for all nodes
    else:
        pos=nx.circular_layout(G)
        
    # nodes
    nx.draw_networkx_nodes(G, pos, ax=ax, node_size=350, nodelist=nself3, 
                           linewidths=3, node_color='orange')
    nx.draw_networkx_nodes(G, pos, ax=ax, node_size=350, nodelist=nself2, 
                           linewidths=2, node_color='lightblue')
    nx.draw_networkx_nodes(G, pos, ax=ax, node_size=350, nodelist=nself1, 
                           linewidths=1, node_color='lightyellow')
    nx.draw_networkx_nodes(G, pos, ax=ax, node_size=350, nodelist=nunself,
                           linewidths=.01, node_color='lightgreen')

    # edges
    nx.draw_networkx_edges(G, pos, ax=ax, edgelist=esmall,
                           width=1, alpha=.5, edge_color='b')
    nx.draw_networkx_edges(G, pos, ax=ax, edgelist=elarge, width=2)
    nx.draw_networkx_edges(G, pos, ax=ax, edgelist=everysmall, width=0.5, style='dotted')
    
    # labels
    nx.draw_networkx_labels(G, pos, ax=ax, font_size=9, font_family='sans-serif')

    ax.axis('off')
    
    return fig
```

```python
def adjacency_matrix(G, nodes=None, disp='weight'):
    
    if nodes is None:
        nodes = sorted(G.nodes())
    
    try:
        Gadj = nx.adjacency_matrix(G, 
                               nodelist=nodes,
                               weight=disp)
        return pd.DataFrame(Gadj.todense(), 
                        index=nodes, 
                        columns=nodes)
    except nx.NetworkXError:
        pass
```

## Search assignment algorithm

```python
def check(GS, GE, n=1, selected_atoms=[], assignments=[]):
    #-----------------------------------------------------
    lines = sorted(GE.nodes())
    atoms = sorted(GS.nodes())
    
    selected_lines=lines[:n]
    GEsub = GE.subgraph(selected_lines)
    GEadj = adjacency_matrix(GEsub)
    print "selected lines: ",selected_lines
    if GEadj is not None:
        pass #display(GEadj)
        
    for atom in atoms:
        nl = n
        assign = selected_atoms[:]
        if atom in assign:
            continue
        assign.append(atom)

        print "check -->", assign
        GSsub = GS.subgraph(assign)
        GSadj = adjacency_matrix(GSsub, assign)
        #display(GEadj)
        #display(GSadj)
        
        if np.all(GSadj.values == GEadj.values):
            #print  selected_lines,"==",assign, "thus continue further"
            # thus we can go further
            nl += 1
            if nl<=len(GE.nodes()):
                final, _ = check(GS, GE, nl, assign, assignments)
                if final is not None:
                    return final
            else:
                print "*"*25
                print 'Possible assignement'
                print assign
                print "*"*25
                assignments.append(assign)
    
    ####
    return None, assignments
    
###
_, res = check(GS, GE)
print "finished - possibles assignements"
GEsub = GE.subgraph(sorted(GE.nodes()))
GEadj = adjacency_matrix(GEsub)
display(GEadj)
for item in res:
    print
    print item
    GSsub = GS.subgraph(item)
    GSadj = adjacency_matrix(GSsub, item)
    display(GSadj)
```
