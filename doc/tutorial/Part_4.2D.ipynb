## Process SER data

The sequence is HMQC with a full echo ajusted with the delay D6 (in s)
```
 ...
(p1 ph1):f1
 d6
(p11 ph11):f3
 d0
(p2 ph2):f1
 d0 
(p11 ph12):f3 
 go=2 ph31
 ...
```
We will thus make a shifted apodization and a rolling of the spectrum   

````python
source = Bruker(path) 
d6 = source.par.D[6]*1e3 # in ms
Process1d(source, transform = 'em 50 --shifted %.3f'%d6)
Process1d(source, transform = 'zf 0; ft')
Process1d(source, transform = 'pk -78 -10')
a = Plot2d(source)
````

* Automatic phasing is also possible

````python
fd = Bruker(path)  # reset to original data
Process1d(fd, transform="em 50; zf 1; ft") # 1 zerofilling
Process1d(fd, "pk --auto")
ax = Plot1d(fd)
````

* Use also phc1 for automatic phasing

````python
fd = Bruker(path)  # reset to original data
Process1d(fd, transform="em 50; zf 1; ft") # 1 zerofilling
Process1d(fd, "pk --auto --fit_phc1 --bound_phc1=180")
a = Plot1d(fd, xlim=(30,-10))
````

* automatic baseline correction

````python
a = Plot1d(fd, hold=True)
Process1d(fd, "ab --mode poly")
a = Plot1d(fd, axe=a)
````

* Show for baseline correction

````python
fd = Bruker(path)  # reset to original data
Process1d(fd, transform="em 50; zf 1; ft; pk --auto --fit_phc1 --bound_phc1=100")
a = Plot1d(fd, hold=True)
Process1d(fd, "ab --mode poly")
a = Plot1d(fd.par.baseline, axe=a, xlim=(30,-10), reverse=True)
````
