## Process STMAS data

** Broadening **

Broadening can take advantage of following the echoes and antiechoes ridges.

With the keywork `--k_shifted` we can do this shifted broadening. p and q must be provided as parameters.

* p is the coherence order: p=3 for 3QMAS, p=1 for STMAS
* q is the satellite order: q=0 for MQMAS, q=2 for first ST-1-MAS, q=4 for ST-2-MAS

Then here we write: `--k_shifted 1 2`

The `k_ratio` give the direction of the ridge.

```python
source = Bruker(path) 
Process(source, transform = 'em 20 --k_shifted 1 2 --verbose')
p_2 = Plot2d(source, figsize=(3,2.7), xlim=(0, 10), ylim=(0, 5))
```

** First Fourier transform **

```python
source = Bruker(path) 
Process(source, transform = 'em 20 --k_shifted 1 2 --verbose')

Process(source, transform = 'zf 1')  # 1 zero-filling
Process(source, transform = 'ft')
Process(source, transform = 'cl -70 0') # clipping
Process(source, transform = 'pk 0 0 --auto --fit_phc1')

p_2 = Plot2d(source, figsize=(3,2.7), ylim=(0, 5))
```

Here we display row 0 and 1

```python
row = source.row(0, byindex=True)
p_1 = Plot1d(row)
```

** Fourier transform in the second dimension (axis=0) **

Note that axis 0 is for F1 and axis 1 is for F2.

```python
source = Bruker(path) 
Process(source, transform = 'em 20 --k_shifted 1 2 --verbose')

# F2 dimension
Process(source, transform = 'zf 1')  # 1 zero-filling
Process(source, transform = 'ft')
Process(source, transform = 'cl -70 0') # clipping
Process(source, transform = 'pk 0 0 --auto --fit_phc1')

# F1 dimension
Process(source, axis=0, transform='em 10')
Process(source, axis=0, transform='zf 1')
Process(source, axis=0, transform='ft')
Process(source, axis=0, transform='pk 0 0 --auto --fit_phc1 --select cols --verbose')

p_2 = Plot2d(source, figsize=(3,2.7), xlim=(-48, -20), ylim=(-40, -20) )
```

** Linear prediction **

We can try to predict missing points at the end of the acquisition in F1

```python
source = Bruker(path) 
Process(source, transform = 'em 20 --k_shifted 1 2 --verbose')

# F2 dimension
Process(source, transform = 'zf 1')  # 1 zero-filling
Process(source, transform = 'ft')
Process(source, transform = 'cl -70 0') # clipping
Process(source, transform = 'pk 0 0 --auto --fit_phc1')

# F1 dimension
Process(source, axis=0, transform='em 10')
Process(source, axis=0, transform='lp 128 --tdeff 106') # must be before any zf command
Process(source, axis=0, transform='zf 1')
Process(source, axis=0, transform='ft')
Process(source, axis=0, transform='pk 0 0 --auto --fit_phc1 --select cols')

p_2 = Plot2d(source, figsize=(3, 2.7), xlim=(-48, -20), ylim=(-40, -20) )
```

** Save this figure **

```python
filename = os.path.join(FIGDIR, 'spec_stmas.png')
p_2.savefig(filename)
```

** Shearing and F1 shifting **

```python
source = Bruker(path) 
#source.par.SF = source.par.BF1


Process(source, transform = 'em 20 --k_shifted 1 2 --verbose')
# Process(source, transform = 'us 3')
# F2 dimension

Process(source, transform = 'zf 2')  # 1 zero-filling
Process(source, transform = 'ft')
Process(source, transform = 'cl -60 0') # clipping
Process(source, transform = 'pk 0 0 --auto --fit_phc1')

# F1 dimension
Process(source, axis=0, transform='em 10')

#Process(source, axis=0, transform='lp 128 --tdeff 106') # must be before any zf command
Process(source, axis=0, transform='zf 1')
Process(source, axis=0, transform='ft --shear 1 2 --shift 1')
Process(source, axis=0, transform='pk 0 0 --auto --fit_phc1 --select cols')

p_2 = Plot2d(source, figsize=(3, 2.7))#, xlim=(-48, -20), ylim=(-40, -20) )
```
