## Fit parameters

```python
%load_ext autoreload
```

```python
%autoreload
from masai.plugins.list import Fit1d, Plot1d
Fit1d = Fit1d()
Plot1d = Plot1d()
```

```python
Fit1d.default()
```

```python
script = """
#-----------------------------------------------------------
# syntax for parameters definition:
# name: value, low_bound,  high_bound
#  * for fixed parameters
#  $ for variable parameters
#  > for reference to a parameter in the COMMON block
#    (> is forbidden in the COMMON block)
# common block parameters should not have a _ in their names
#-----------------------------------------------------------
#

COMMON:
# common parameters ex.
# $ gwidth: 1.0, 0.0, none
  $ gratio: 0.5, 0.0, 1.0

MODEL: LINE_1
shape: voigtmodel
    * ampl:  1.0, 0.0, none
    $ pos:   4.5, -100.0, 100.0
    > ratio: gratio
    $ width: 4.0, 0, 100
    
MODEL: LINE_2
shape: voigtmodel
    $ ampl:  0.2, 0.0, none
    $ pos:   1.6, -100.0, 100.0
    > ratio: gratio
    $ width: 4.0, 0, 100

MODEL: LINE_3
shape: voigtmodel
    $ ampl:  0.18, 0.0, none
    $ pos:   7.9, -100.0, 100.0
    > ratio: gratio
    $ width: 4.0, 0, 100

MODEL: LINE_4
shape: voigtmodel
    $ ampl:  0.1, 0.0, none
    $ pos:   14, 10.0, 20.0
    > ratio: gratio
    $ width: 6.0, 0, 100

MODEL: LINE_5
shape: voigtmodel
    $ ampl:  0.1, 0.0, none
    $ pos:   -5, -8.0, -1.0
    > ratio: gratio
    $ width: 6.0, 0, 100
"""
f = Fit1d(source, script)
ax = Plot1d(source, 
            figsize=(3.4, 2.0), dpi=180, 
            xlim=(30, -20), lw=.5,
            plot_model=True,
            offset=5)
```

```python
f.run()
ax = Plot1d(source, 
            figsize=(3.4, 2.0), dpi=180, 
            xlim=(30, -20), lw=.5,
            plot_model=True,
            offset=5)
```
